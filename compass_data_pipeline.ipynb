```python
import yfinance as yf
import pandas as pd
compass = yf.Ticker("CPG.L")
pd.set_option('display.float_format', lambda x: '%.2f' % x)

income_statement = compass.financials
income_statement_3yr = income_statement.iloc[:, :3]
print("--- Income Statement ---")
display(income_statement_3yr)

balance_sheet = compass.balance_sheet
balance_sheet_3yr = balance_sheet.iloc[:, :3]
print("--- Balance Sheet ---")
display(balance_sheet_3yr)

cash_flow = compass.cashflow
cash_flow_3yr = cash_flow.iloc[:, :3]
print("--- Cash Flow Statement ---")
display(cash_flow_3yr)
```

```python
income_statement_3yr.to_csv("compass_income_statement_3yr.csv")
balance_sheet_3yr.to_csv("compass_balance_sheet_3yr.csv")
cash_flow_3yr.to_csv("compass_cash_flow_3yr.csv")
```
